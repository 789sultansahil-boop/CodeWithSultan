**Q.6 Create a 𝐟𝐮𝐧𝐜𝐭𝐢𝐨𝐧 multiply_list(lst, factor).**

```python
def multiply_list(lst, factor):
    return [x * factor for x in lst]

test_list = [1, 2, 3]
print(multiply_list(test_list, 2))
```

**Q.7 Use 𝐥𝐞𝐧() and write 𝐟𝐮𝐧𝐜𝐭𝐢𝐨𝐧 length_info(x).**

```python
def length_info(x):
    length = len(x)
    return f"The length of {x} is {length}"

print(length_info("Hello"))
print(length_info([1, 2, 3]))
```

**Q.9 Write 𝐟𝐮𝐧𝐜𝐭𝐢𝐨𝐧 circle_area(radius).**

```python
import math

def circle_area(radius):
    return math.pi * radius * radius

print(circle_area(2))
```
